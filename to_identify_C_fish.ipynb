```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# loading data
data = np.load('/home/hck03/Documents/MS_thesis/tracking_output/1c_4n_faulty/Clip0095_E2_7/data/clean_tracks.npy')
print(data.shape)
```

```python
# finding kinematic variables
## distance-time variables
vel = np.gradient(data, axis=1)
speed = np.linalg.norm(vel, axis=2)
acc_mag = np.gradient(speed, axis=1)

## manoeuvrability variables
heading_vec = vel/speed[:,:,None]
heading_angles = np.arctan2(heading_vec[:,:,1], heading_vec[:,:,0])

angular_vel = np.gradient(heading_angles, axis=1)
```
